## imports

```python
import sys, os
from pathlib import Path

# Ruta absoluta a la raíz del proyecto (donde están Entrega1, Entrega2, Entrega3)
ROOT = Path().resolve().parent  

while not (ROOT / "Entrega3").exists():
    ROOT = ROOT.parent

print("ROOT DETECTED:", ROOT)

sys.path.append(str(ROOT))
sys.path.append(str(ROOT / "Entrega3"))
sys.path.append(str(ROOT / "Entrega3" / "src"))

from Entrega3.src.models.load_artifacts import load_model_artifacts
from Entrega3.src.utils.preprocessing import frames_to_feature_vector
from Entrega3.src.online.realtime_inference import RealtimeHARPredictor

```

## Prueba sintética

```python
frames = []
for i in range(60):
    frames.append({
        "video_id": "synthetic",
        "frame_index": i,
        "timestamp": i/30,
        "t_start": i/30,
        "t_end": (i+1)/30,
        "landmarks": {
            k: {"x": 0.5, "y": 0.5, "z": 0, "visibility": 1.0}
            for k in ["left_shoulder","right_shoulder","left_hip","right_hip",
                      "left_knee","right_knee","left_ankle","right_ankle",
                      "left_wrist","right_wrist","left_ear","right_ear","head"]
        }
    })
```

## Prueba del predictor:

```python
pred = RealtimeHARPredictor(fps=30)

for fr in frames:
    pred.add_frame(fr["landmarks"])

print(pred.maybe_predict())

```
